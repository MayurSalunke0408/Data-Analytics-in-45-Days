```python
import pandas as pd

# Sample data
data = pd.DataFrame({
    'drink': ['Coke', 'Latte', 'Orange Juice', 'Green Tea', 'Espresso', 'Pepsi']
})

# Dictionary mapping
drink_to_category = {
    'coke': 'soft drink',
    'pepsi': 'soft drink',
    'latte': 'coffee',
    'espresso': 'coffee',
    'green tea': 'tea',
    'orange juice': 'juice'
}

# Function approach
def drink_to_type(series):
    if series['drink'] == 'coke':
        return 'soft drink.'
    elif series['drink'] == 'pepsi':
        return 'soft drink.'
    elif series['drink'] == 'latte':
        return 'coffee'
    elif series['drink'] == 'espresso':
        return 'coffee'
    elif series['drink'] == 'green tea':
        return 'tea'
    else:
        return 'juice'


# Create new column using dictionary mapping
data['category'] = data['drink'].map(str.lower).map(drink_to_category)

print(data)
```

We frequently find missing values in our data set. A quick method for imputing missing values is by filling the missing value with any random number. Not just missing values, you may find lots of outliers in your data set, which might require replacing. Let's see how can we replace values.

```python
#Series function from pandas are used to create arrays
data = pd.Series([1., -999., 2., -999., -1000., 3.])
data
```

```python
#replace -999 with NaN values
data.replace(-999, np.nan,inplace=True)
data
```

```python
#We can also replace multiple values at once.
data = pd.Series([1., -999., 2., -999., -1000., 3.])
data.replace([-999,-1000],np.nan,inplace=True)
data
```

```python
# how to rename column names and axis (row names).
data = pd.DataFrame(np.arange(12).reshape((3, 4)),index=['Ohio', 'Colorado', 'New York'],columns=['one', 'two', 'three', 'four'])
data
```

```python
#Using rename function
data.rename(index = {'Ohio':'SanF'}, columns={'one':'one_p','two':'two_p'},inplace=True)
data
```

```python
#You can also use string functions
data.rename(index = str.upper, columns=str.title,inplace=True)
data
```
